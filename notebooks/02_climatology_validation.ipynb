# Vajra — 02: Climatology & Extreme Forecast Index (EFI) Validation
### MoES / NCMRWF Problem Statement 26078
This notebook computes the 30-year climatological percentiles (P90, P95, P99) and calculates the Extreme Forecast Index (EFI ∈ [-1, 1]) to isolate anomalous atmospheric states.

```python
import numpy as np
import matplotlib.pyplot as plt
from vajra.tracking.efi_calculator import EFICalculator, make_synthetic_anomaly_fields

# Generate synthetic forecast ensemble and climatology baseline (Bay of Bengal cyclone footprint)
fcst_dict, clim_dict = make_synthetic_anomaly_fields(H=60, W=80, event_type="cyclone", seed=42)
calc = EFICalculator(n_percentile_points=51, efi_threshold=0.70)

results = calc.compute_all_variables(fcst_dict, clim_dict)
print("Evaluated variables:", list(results.keys()))
```

```python
# Plot Precipitation EFI Field and Anomaly Trigger Mask
efi_tp = results["tp"]["efi"]
zscore_tp = results["tp"]["zscore"]
mask_tp = results["tp"]["anomaly_mask"]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
im0 = axes[0].imshow(efi_tp, cmap="coolwarm", vmin=-1, vmax=1)
axes[0].set_title("Total Precipitation EFI ([-1, 1])")
plt.colorbar(im0, ax=axes[0])

im1 = axes[1].imshow(zscore_tp, cmap="YlOrRd")
axes[1].set_title("Z-Score Anomaly")
plt.colorbar(im1, ax=axes[1])

im2 = axes[2].imshow(mask_tp, cmap="Blues")
axes[2].set_title("Triggered Anomaly Mask (|EFI| >= 0.70)")
plt.colorbar(im2, ax=axes[2])
plt.tight_layout()
plt.show()
```

```python
# Compute multi-variable composite anomaly
composite = calc.composite_anomaly_flag(results, event_type="cyclone")
print(f"Composite anomalous grid cells: {composite.sum()} / {composite.size}")
```
